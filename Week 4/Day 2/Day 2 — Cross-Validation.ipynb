```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import f1_score
```

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

df.head()
```

```python
df["Partner"] = df["Partner"].fillna(df["Partner"].mode()[0])
df["MonthlyCharges"] = df["MonthlyCharges"].fillna(df["MonthlyCharges"].median())
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df["TotalCharges"] = df["TotalCharges"].fillna(df["TotalCharges"].median())
```

```python
df["gender"] = df["gender"].replace({ "Male" : 1, "Female" : 0})
df["Partner"] = df["Partner"].replace({ "Yes" : 1, "No" : 0})
df["Dependents"] = df["Dependents"].replace({ "Yes" : 1, "No" : 0})
df["PhoneService"] = df["PhoneService"].replace({ "Yes" : 1, "No" : 0})
df["MultipleLines"] = df["MultipleLines"].replace({ "Yes" : 1, "No" : 0, "No phone service" : 0 })
df["OnlineSecurity"] = df["OnlineSecurity"].replace({ "Yes" : 1, "No" : 0, "No internet service" : 0 })
df["OnlineBackup"] = df["OnlineBackup"].replace({ "Yes" : 1, "No" : 0, "No internet service" : 0 })
df["DeviceProtection"] = df["DeviceProtection"].replace({ "Yes" : 1, "No" : 0, "No internet service" : 0 })
df["TechSupport"] = df["TechSupport"].replace({ "Yes" : 1, "No" : 0, "No internet service" : 0 })
df["StreamingTV"] = df["StreamingTV"].replace({ "Yes" : 1, "No" : 0, "No internet service" : 0 })
df["StreamingMovies"] = df["StreamingMovies"].replace({ "Yes" : 1, "No" : 0, "No internet service" : 0 })
df["PaperlessBilling"] = df["PaperlessBilling"].replace({ "Yes" : 1, "No" : 0 })
df["StreamingTV"] = df["StreamingTV"].replace({ "Yes" : 1, "No" : 0, "No internet service" : 0 })
df["Churn"] = df["Churn"].replace({"Yes": 1, "No": 0})
df = pd.get_dummies(df, columns=["InternetService", "Contract", "PaymentMethod"])
```

```python
X = df.drop(columns=["customerID", "Churn"])
Y = df["Churn"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()

X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train),
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    scaler.transform(X_test),
    columns=X_test.columns,
    index=X_test.index
)

X_train_scaled.head()
```

Step 1: Take a Week 3 model and evaluate it with 5-fold cross-validation using cross_val_score.

```python
model1 = LogisticRegression(C=1, max_iter=1000, class_weight='balanced', random_state=42)
scores = cross_val_score(model1, X_train_scaled, Y_train, cv=5, scoring="f1")
print(scores)
```

Step 2: Report the mean and standard deviation of the scores across folds.

```python
print("Mean =",scores.mean())
print("Std =",scores.std()) 
```

we can see that 5 scores f1 = [0.648, 0.621, 0.613, 0.610, 0.615] roughly the same and we have a small std, in this case we can say that the model is stable and that's a good sign, so we can trust the 0.622 scores.

Step 3: Compare the cross-validated estimate to the single-split score from Day 1 and explain any 
difference.

```python
X_train2, X_val, Y_train2, Y_val = train_test_split(
    X_train, Y_train, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()

X_train2_scaled = pd.DataFrame(
    scaler.fit_transform(X_train2),
    columns=X_train2.columns,
    index=X_train2.index
)

X_val_scaled = pd.DataFrame(
    scaler.transform(X_val),
    columns=X_val.columns,
    index=X_val.index
)

X_train_scaled.head()
```

```python
best_f1 = 0
best_C = None

for C in [0.1, 1, 10, 100]:
    model2 = LogisticRegression(C=C, max_iter=1000, class_weight='balanced', random_state=42)
    model2.fit(X_train2_scaled, Y_train2)
    pred = model2.predict(X_val_scaled)
    f1 = f1_score(Y_val, pred, pos_label=1)
    print(f"C = {C}, f1 score = {f1:.4f}")
    if f1 > best_f1:
        best_f1 = f1
        best_C = C

print(f"the best C = {best_C}, with f1 score = {best_f1:.4f}")
```

the single train/val split gives us a higher f1 (0.6309) than thwe 5-fold cross-val avg (0.6218). in single train/val split the model could get a bit lucky. since the cross-val std is small (0.0135), this difference is just normal variation. I trust the 5-fold cross-val result (0.6218) more, since it's the more reliable estimate.

Step 4: For a classification task, confirm stratified folds are used and explain why that matters here.

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores_skf = cross_val_score(model1, X_train_scaled, Y_train, cv=skf, scoring="f1")
print(scores_skf)
```

```python
print("Mean =",scores_skf.mean())
```

I ran StratifiedKFold and got these scores = [0.5897, 0.6243, 0.6245, 0.6175, 0.6534], scores are very close to the scores from the default cross_val_score(...) used in step1, whoch confirms that cross_val_score already used stratified folds automatically when I passed an integer cv=5 with classifier.
why that matters: this is an imbalanced churn dataset non-stratified folds could end up with very different churn ratios by chance, making the f1 scores unreliable to compare. stratified k-fold keeps the same churn ratioin every fold, so the score is fair, trustworthy estimate.
