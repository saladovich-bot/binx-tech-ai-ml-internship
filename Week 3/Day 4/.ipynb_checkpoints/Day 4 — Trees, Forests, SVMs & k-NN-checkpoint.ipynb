```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv("train.csv")

df.head()
```

```python
df.info()
```

```python
df = df.drop(columns=["Cabin"])
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df["Sex"] = df["Sex"].replace({"male" : 1, "female" : 0})
df = pd.get_dummies(df, columns= ["Embarked"])
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
X = df.drop(columns=["Survived", "Name", "Ticket"], axis=1)
Y = df["Survived"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42)
```

Step 1: Train a decision tree, random forest, SVM, and k-NN on the same train/test split from Day 3.

```python
model1 = DecisionTreeClassifier(max_depth=5, random_state=42)
model1.fit(X_train,Y_train)
pred1 = model1.predict(X_test)
```

```python
model2 = RandomForestClassifier(n_estimators=100, random_state=42)
model2.fit(X_train,Y_train)
pred2 = model2.predict(X_test)
```

```python
model3 = SVC(kernel="rbf", probability=True)
model3.fit(X_train,Y_train)
pred3 = model3.predict(X_test)
```

```python
model4 = KNeighborsClassifier(n_neighbors=5)
model4.fit(X_train,Y_train)
pred4 = model4.predict(X_test)
```

Step 2: Evaluate all four with the same metric (e.g. F1-score) and assemble the results into one comparison 
table.

```python
print(classification_report(Y_test, pred1))
```

```python
print(classification_report(Y_test, pred2))
```

```python
print(classification_report(Y_test, pred3))
```

```python
print(classification_report(Y_test, pred4))
```

```python
comparison_table = pd.DataFrame({
    "Model" : ["Decision Tree", "Random Forest", "SVC", "KNeighbors"],
    "F1-score" : [0.80, 0.83, 0.48, 0.64]
    })
```

```python
comparison_table.head()
```

here we look at the F1-score for all 4 models to pick the best one by comparing them. we can see that Random Forest = 0.83 performed the best 'Cause it achieved the highest F1-score.

Step 3: Report the random forest's top feature importances and interpret them.

```python
importances_table = pd.DataFrame({
    "feature" : X_train.columns,
    "importances" : model2.feature_importances_
    })
```

```python
importances_table = importances_table.sort_values(
    by="importances",
    ascending=False
)
```

```python
print(importances_table)
```

we can see in this table the importances of each features in our dataset and it's effect on prediction, Sex have the highest importances meaning that this feature is very important to get our prediction from the model.
in the other hand we have Embarked with the lowest prediction, not very important to our model. we can experement later and remove it from the features.
