```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
df = pd.read_csv("house_price_regression_dataset.csv")

df.head()
```

```python
X = df.drop("House_Price", axis=1)
Y = df["House_Price"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42)
```

Step 1: Train a LinearRegression model on a provided regression dataset (e.g. house prices).

```python
model = LinearRegression()
model.fit(X_train, Y_train)
predictions = model.predict(X_test)
```

Step 2: Report the model's coefficients and identify which feature has the strongest effect.

```python
print(model.coef_)
print(model.intercept_)
```

the strongest coefficient is the largest one, Lot_Size has the strongest effect since it has a largest coefficient (14885.38441462), it means that one-unit increase in Lot_Size is the highest increase in House_Price.
but in fact I want to add a note from me, it is not mean that it is the strongest feature, 'cause the numbers in Lot_Size is too small if we compare it for example with Square_Footage.

Step 3: Evaluate the model with MAE, RMSE, and R² on the test set.

```python
print("MAE =", mean_absolute_error(Y_test, predictions))
print("RMSE =", np.sqrt(mean_squared_error(Y_test, predictions)))
print("R² =", r2_score(Y_test, predictions))
```

Step 4: Compare the RMSE against a baseline that predicts the mean for every row, and state whether the 
model adds value.

```python
baseline_pred = np.full_like(Y_test, Y_train.mean(), dtype=float)
baseline_rmse = np.sqrt(mean_squared_error(Y_test, baseline_pred))
print(baseline_rmse)
print(np.sqrt(mean_squared_error(Y_test, predictions)))
```

Baseline RMSE (predicting the mean for every row) = 253891.4870598681, and the model RMSE = 10071.484424138673. the model RMSE is smaller than the baseline, and R2 = 0.998, this means that the model explains 99.8% of the variance in house price. it is very good.
