```python
import numpy as np
import pandas as pd
```

 Load a numeric column from a provided dataset into a NumPy array or Pandas Series:

```python
df = pd.read_csv("data.csv")

df.head(10)
```

```python
df.isnull().sum()
```

```python
df["Calories"] = df["Calories"].fillna(df["Calories"].median())
```

```python
df.drop_duplicates(inplace=True)
```

```python
nums = np.array(df["Calories"])
```

```python
print(nums)
```

Compute its mean, median, mode, standard deviation, and IQR:

```python
mean = np.mean(nums)
median = np.median(nums)
mode = df["Calories"].mode()[0]
```

```python
std = np.std(nums)
q1, q3 = np.percentile(nums, [25,75])
IQR = q3 - q1
```

```python
print("Mean =", mean)
print("Median =", median)
print("mode =", mode)
print("std =", std)
print("IQR =", IQR)
```
