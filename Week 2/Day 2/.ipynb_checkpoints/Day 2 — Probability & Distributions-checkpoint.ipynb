 Simulate 10,000 coin flips with NumPy and confirm the proportion of heads approaches 0.5 and Document each result with a Markdown explanation of what it demonstrates:

```python
flips = np.random.choice(['heads', 'tails'], size=10000)
prop_heads = np.sum(flips == 'heads') / len(flips)
print(prop_heads)
```

Given a small worked scenario, compute a conditional probability by hand and verify it with a 
simulation:

suppose we have 5 cards, 3 red and 2 blue, if we got a red card at first, whats the probability of gitting a red card then?

```python
cards = ["red"]*3 + ["blue"]*2
count = 0
red_first = 0
trails = 10000

for _ in range(trails):
    draw = np.random.choice(cards, size=2,replace=False)
    if draw[0] == "red":
        red_first += 1
        if draw[1] == "red":
            count += 1

print(count/red_first)
```

I create a list of 5 cards( 3 - red, 2 - blue), count to calculate how many red cards we will get in the second choice, and re_first to be sure that the first card will be red as we want. then the loop will loop 10000. using np.random.choice we get two choice of cards in random way (red or blue), if the first card is red (we +1 to the red_first) the program goes to the second condition when the second card is also red (+1 to the count). then we calculate the probability of appering red card in the second choice when the first is red. we got ~0.5, 'cause in the second choice there is only two choices(red or blue).
