 Apply the IQR method in code to flag outliers in at least one column, and decide (with justification) 
how to handle them:

```python
Q1 = df["Pulse"].quantile(0.25)
Q3 = df["Pulse"].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df["Pulse"] < Q1 - 1.5*IQR) | (df["Pulse"] > Q3 + 1.5*IQR)]

outliers.head
```

```python
Q1 = df["Maxpulse"].quantile(0.25)
Q3 = df["Maxpulse"].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df["Maxpulse"] < Q1 - 1.5*IQR) | (df["Maxpulse"] > Q3 + 1.5*IQR)]

print(outliers)
```

we detect a few outliers here in pulse and maxpulse, I will keep them 'Cause it will affect the prediction since there are a few of them.
