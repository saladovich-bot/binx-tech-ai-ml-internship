in the first cell I define a func. that takes a list of numbers, then I create a dict(calc) to store in it 
what we want, then I calculate the mean (which means avg), min (i use buildin func.) and the max,
and I store every thing in the dic calc and return it.
then I create a list with numbers from 1 to 30, and call my func.

```python
def meanminmax(l):
    calc = {}
    calc["mean"] = sum(l)/len(l)
    calc["min"] = min(l)
    calc["max"] = max(l)
    return calc
```

```python
mylist = [26, 7, 30, 11]
print(meanminmax(mylist))
```

I create a class named GPA_pass, then inside it I define a __init__ func. to pass the parameters, so the class can use them.
word self tell us that these parameters belongs to the class GPA_pass. then I define first method called pass_or_fail that tell us if the student pass the course or fail ,using if statement if the gpa greater than 0.7 - pass, else the student fail.

```python
class GPA_pass:
    def __init__(self, course_name, stu_name, gpa):
        self.course_name = course_name
        self.stu_name = stu_name
        self.gpa = gpa

    def pass_or_fail(self):
        if self.gpa > 0.7:
            print(f"{self.course_name} - {self.stu_name} : pass")
        else:
            print(f"{self.course_name} - {self.stu_name} : fail")
```

```python
student1 = GPA_pass("python", "Sami", 3.33)
student2 = GPA_pass("python", "Marah", 0.7)

student1.pass_or_fail()
student2.pass_or_fail()
```
