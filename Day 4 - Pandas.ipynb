Count missing values per column and handle them (fill or drop) with a documented justification:

```python
print(df.isnull().sum())
```

justification: we have two columns that contains missing values, I replace missing valeus in Age, and delete column Cabin 'Cause there is too many missing valuse (687/891), it may cause problems in prediction.

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df.isnull().sum())
```

```python
df = df.drop(columns=["Cabin"])
print(df.isnull().sum())
```
